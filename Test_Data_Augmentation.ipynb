# Test Data Augmentation Using OpenAI API

## Overview
Automatically generate diverse test cases and question variants from seed data using LLM-powered augmentation.

---

## Why Augment Test Data?
- **Expand test coverage** without manual effort
- **Generate realistic variants** of questions and edge cases
- **Reduce bias** in test data by exploring diverse phrasings
- **Scale testing** for NLP and Q&A systems (e.g., NotebookLM evaluation)

---

## Our Implementation: `augment_questions.py`

### Input
```
artifacts/queries.txt (or semantic_results.csv)
List three primary colors in traditional color theory.
What is test complexity?
Name two advantages of automated software testing.
```

### Process
1. **OpenAI Mode** (if `OPENAI_API_KEY` set): Call GPT-3.5-turbo / GPT-4 to generate semantic paraphrases
2. **Fallback Mode** (no API key): Use deterministic templates to generate local variants

### Output
```json
{
  "timestamp": "2025-12-06T07:30:04.851244+00:00",
  "original": "List three primary colors in traditional color theory.",
  "augmented": "What are the primary answers to: List three primary colors in traditional color theory?",
  "method": "openai"
}
```
→ Saved to `artifacts/augmented_questions.jsonl` (9 variants from 3 queries; configurable via `--n`)

---

## Quick Usage

### With OpenAI API Key
```bash
export OPENAI_API_KEY='sk-...'
python scripts/augment_questions.py \
  --queries-file artifacts/queries.txt \
  --out artifacts/augmented_questions.jsonl \
  --n 5 --model gpt-3.5-turbo
```

### Without API Key (Fallback)
```bash
python scripts/augment_questions.py \
  --queries-file artifacts/queries.txt \
  --out artifacts/augmented_questions.jsonl \
  --n 3
```

## Example: Semantic Similarity Testing
**Use case**: Validate NotebookLM responses across multiple question phrasings

1. Run selenium test with original query → get expected/actual text → compute similarity score
2. **Augment** original query using this script → get 5 variants
3. **Re-run** tests with augmented queries (manual or via pytest parametrization)
4. **Aggregate** all semantic scores in CSV → report consistency across phrasings

Result: Confidence that the system answers consistently regardless of how a question is phrased.
